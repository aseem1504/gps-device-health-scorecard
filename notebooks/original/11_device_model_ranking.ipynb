```python
device_performance=pd.read_sql("select * from device_performance limit 10000",con=postgres)
```

```python
model_names=pd.read_sql("select * from device_models",con=mysql_cn)
```

```python
device_performance.columns
```

```python
data=[]
for i,row in model_names.iterrows():
    a=[]
    device_specs=device_performance[device_performance['model_name']==row['name']]
    a.append(row['name'])
    a.append(device_specs.averagedelay.mean())
    a.append(device_specs.jumpdistance.mean())
    a.append(device_specs.usedsat.mean())
    a.append(device_specs.connected_percent.mean())
    a.append(device_specs.jump_percent.mean())
    a.append(device_specs.drivingtime.mean()/device_specs.drivingrawcount.mean())
    a.append(device_specs.stoppagetime.mean()/device_specs.stoppagerawcount.mean())
    data.insert(0,a)
    
    
    
    
```

```python
table=pd.DataFrame(data,columns=['model_name','averagedelay','jumpdistance','usedsat','connected_percent','jump_percent','driving_ppings_fequency','stoppage_ping_frequency'])
```
